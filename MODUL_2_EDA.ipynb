```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())        # 6300.0
print('Harga tertinggi:', harga.max())          # 12000
print('Harga setelah diskon 10%:', harga * 0.9) # [4500. 6300. 2700. 10800. 4050.]
```

```python
import pandas as pd
df = pd.read_csv('data_kantin.csv')
print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

```python
print(df.isnull().sum())              # menu: 3, terjual: 6
df['terjual'] = df['terjual'].fillna(0)      # isi kekosongan dengan 0
df = df.dropna(subset=['menu'])              # hapus baris tanpa nama menu
```

```python
print(df.duplicated().sum())   # 5 baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int)
```

```python
laris = df[df['terjual'] > 20]                         # 47 baris menu laris
urut = df.sort_values(by='terjual', ascending=False)   # Es Teh 45x jadi teratas
df['total_pendapatan'] = df['harga'] * df['terjual']
ringkasan = df.groupby('menu')['total_pendapatan'].sum().sort_values(ascending=False)
```
